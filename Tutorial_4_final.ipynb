```python
# Imports
import numpy as np
import pandas as pd
import statsmodels.api as sm

from scipy.stats import chi2_contingency

pd.set_option("display.max_columns", None)
```

## 1. Multiple Linear Regression

How well do Age, Income, Household_Size and Gender explain Average_Spend? What are the underlying assumptions when doing a linear regression analysis?

```python
# Load Dataset
df = pd.read_csv("")
```

```python
# Create a copy of the DataFrame and encode gender as a binary variable:
reg_data = (
    df.assign(Male=lambda d: (d["Gender"] == "Male").astype(int))
)

# Add a constant term to estimate the intercept and select the independent variables
X = sm.add_constant(
    reg_data[["Age", "Income", "Household_Size", "Male"]].astype(float)
)

# Define the dependent variable
y = reg_data["Average_Spend"]

# Fit a multiple linear regression model using Ordinary Least Squares (OLS)
multiple_model = sm.OLS(y, X).fit()

# Output
print("\nMultiple Linear Regression")
print("-----------------------------")
print(f"n = {int(multiple_model.nobs)}")
print(multiple_model.summary())
```

## 2. Chi-Square Test of Independence

**Is there an association between Education_Level and Marital_Status?**

```python
# Create a crosstab
crosstab = pd.crosstab(df["Education_Level"], df["Marital_Status"])

# Output
print("\nCrosstab: Education Level x Marital Status")
print("-----------------------------")
print(crosstab)

# Perform a chi-square test of independence to determine whether the two categorical variables are associated
chi2, p, dof, _ = chi2_contingency(crosstab)

# Calculate Cramér's V to measure the strength of the association between the two variables
cramers_v = np.sqrt(
    chi2 / (crosstab.to_numpy().sum() * (min(crosstab.shape) - 1))
)

# Output
print("\nChi-Square Test of Independence")
print("-----------------------------")
print(f"Chi-square:   {chi2:.3f}")
print(f"df:           {dof}")
print(f"p = {p:.3f}")
print(f"Cramér's V:   {cramers_v:.3f}")
```

If 𝑝 < 0,05 → Reject the Null

If 𝑝 > 0,05 → Cannot reject the Null
